```python
before_duplicates = len(df)

df = df.drop_duplicates(keep="first").copy()

after_duplicates = len(df)

print("Rows before:", before_duplicates)
print("Rows after:", after_duplicates)
print("Duplicates removed:", before_duplicates - after_duplicates)
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.replace(r"\s+", "_", regex=True)
    .str.replace(r"[^A-Za-z0-9_]", "", regex=True)
)

print(df.columns.tolist())
```

```python
text_columns = df.select_dtypes(include=["object", "string"]).columns

for col in text_columns:
    df[col] = df[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

print("Extra spaces removed.")
```

```python
categorical_columns = [
    "Product_Category",
    "Order_Status",
    "Payment_Method",
    "Sales_Channel",
    "City",
    "State"
]

for col in categorical_columns:
    df[col] = (
        df[col]
        .astype("string")
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
        .str.title()
    )

df["Coupon_Code"] = (
    df["Coupon_Code"]
    .astype("string")
    .str.strip()
    .str.upper()
)

df["Payment_Method"] = df["Payment_Method"].replace({
    "Cash On Delivery": "Cash on Delivery"
})

print("Text standardization completed.")
```

```python
print("Categories:")
display(df["Product_Category"].value_counts(dropna=False))

print("Order Status:")
display(df["Order_Status"].value_counts(dropna=False))

print("Payment Method:")
display(df["Payment_Method"].value_counts(dropna=False))
```

```python
numeric_columns = [
    "Quantity",
    "Unit_Price_INR",
    "Discount_Percent",
    "Sales_Amount_INR",
    "Unit_Cost_INR",
    "Total_Cost_INR",
    "Profit_INR",
    "Shipping_Fee_INR",
    "Customer_Rating"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print(df[numeric_columns].dtypes)
```

```python
invalid_quantity = df[
    df["Quantity"].isna() | (df["Quantity"] <= 0)
]

print("Invalid quantity records:", len(invalid_quantity))

display(invalid_quantity)
```

```python
invalid_discount = df[
    df["Discount_Percent"].isna()
    | ~df["Discount_Percent"].between(0, 100)
]

print("Invalid discount records:", len(invalid_discount))

display(invalid_discount)
```

```python
invalid_ratings = df[
    df["Customer_Rating"].notna()
    & ~df["Customer_Rating"].between(1, 5)
]

print("Invalid ratings:", len(invalid_ratings))

display(invalid_ratings)
```

```python
date_columns = [
    "Order_Date",
    "Ship_Date",
    "Delivery_Date"
]

for col in date_columns:
    df[col] = pd.to_datetime(
        df[col],
        format="mixed",
        dayfirst=True,
        errors="coerce"
    )

print(df[date_columns].dtypes)
```

```python
for col in date_columns:
    print(col, "missing dates:", df[col].isna().sum())
```

```python
invalid_ship_date = (
    df["Ship_Date"].notna()
    & (df["Ship_Date"] < df["Order_Date"])
)

invalid_delivery_date = (
    df["Delivery_Date"].notna()
    & (df["Delivery_Date"] < df["Order_Date"])
)

print("Ship date before order:", invalid_ship_date.sum())
print("Delivery date before order:", invalid_delivery_date.sum())
```

```python
invalid_mask = (
    df["Quantity"].isna()
    | (df["Quantity"] <= 0)
    | df["Discount_Percent"].isna()
    | ~df["Discount_Percent"].between(0, 100)
    | df["Unit_Price_INR"].isna()
    | (df["Unit_Price_INR"] < 0)
    | df["Unit_Cost_INR"].isna()
    | (df["Unit_Cost_INR"] < 0)
    | df["Shipping_Fee_INR"].isna()
    | (df["Shipping_Fee_INR"] < 0)
)

invalid_transactions = df[invalid_mask].copy()

invalid_transactions.to_csv(
    "ecommerce_invalid_transactions_review.csv",
    index=False
)

print("Invalid records saved:", len(invalid_transactions))
```

```python
df = df.loc[~invalid_mask].copy()

print("Rows remaining after validation:", len(df))
```

```python
text_fill_columns = [
    "Customer_Name",
    "Product_Category",
    "Payment_Method",
    "City",
    "State"
]

for col in text_fill_columns:
    df[col] = df[col].fillna("Unknown")

df["Coupon_Code"] = df["Coupon_Code"].fillna("UNKNOWN")

print("Descriptive missing values handled.")
```

```python
invalid_rating_mask = (
    df["Customer_Rating"].notna()
    & ~df["Customer_Rating"].between(1, 5)
)

df.loc[invalid_rating_mask, "Customer_Rating"] = np.nan

print("Invalid ratings converted to missing values.")
```

```python
df["Sales_Amount_INR"] = (
    df["Quantity"]
    * df["Unit_Price_INR"]
    * (1 - df["Discount_Percent"] / 100)
    + df["Shipping_Fee_INR"]
).round(2)

print("Sales amount recalculated.")
```

```python
df["Total_Cost_INR"] = (
    df["Quantity"] * df["Unit_Cost_INR"]
    + df["Shipping_Fee_INR"] * 0.55
).round(2)

print("Total cost recalculated.")
```

```python
df["Profit_INR"] = (
    df["Sales_Amount_INR"] - df["Total_Cost_INR"]
).round(2)

df["Profit_Margin_Percent"] = np.where(
    df["Sales_Amount_INR"] != 0,
    df["Profit_INR"] / df["Sales_Amount_INR"] * 100,
    np.nan
).round(2)

display(
    df[
        [
            "Order_ID",
            "Sales_Amount_INR",
            "Total_Cost_INR",
            "Profit_INR",
            "Profit_Margin_Percent"
        ]
    ].head()
)
```

```python
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate Order IDs:", df["Order_ID"].duplicated().sum())
print("Final shape:", df.shape)
```

```python
print("Invalid quantity:", (df["Quantity"] <= 0).sum())

print(
    "Invalid discount:",
    (~df["Discount_Percent"].between(0, 100)).sum()
)

print(
    "Invalid ratings:",
    (
        df["Customer_Rating"].notna()
        & ~df["Customer_Rating"].between(1, 5)
    ).sum()
)

print(
    "Profit calculation mismatch:",
    (
        (df["Profit_INR"] - (
            df["Sales_Amount_INR"] - df["Total_Cost_INR"]
        )).abs() > 0.01
    ).sum()
)
```

```python
final_missing_report = pd.DataFrame({
    "Missing_Count": df.isna().sum(),
    "Missing_Percentage": (df.isna().mean() * 100).round(2)
})

display(
    final_missing_report.sort_values(
        "Missing_Percentage",
        ascending=False
    )
)
```

```python
# Check missing values in every column

missing_count = df.isnull().sum()

missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_report = pd.DataFrame({
    "Missing_Count": missing_count,
    "Missing_Percentage": missing_percentage.round(2)
})

# Missing values irukkura columns mattum kaattum
missing_report = missing_report[
    missing_report["Missing_Count"] > 0
].sort_values(by="Missing_Count", ascending=False)

print("Missing Values - Recheck Report")
display(missing_report)
```

```python
# Keep missing ratings as missing
df["Customer_Rating"] = pd.to_numeric(
    df["Customer_Rating"], errors="coerce"
)

# Keep date columns in datetime format
df["Ship_Date"] = pd.to_datetime(
    df["Ship_Date"], errors="coerce"
)

df["Delivery_Date"] = pd.to_datetime(
    df["Delivery_Date"], errors="coerce"
)

print("Columns converted successfully!")
```
